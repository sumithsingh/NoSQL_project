To create patient node

```python
def create_patient(tx, patient_id, name):
    query = """
    CREATE (p:Patient {id: $patient_id, name: $name})
    """
    tx.run(query, patient_id=patient_id, name=name)
```

To create doctor node

```python
def create_doctor(tx, doctor_id, name, specialty):
    query = """
    CREATE (d:Doctor {id: $doctor_id, name: $name, specialty: $specialty})
    """
    tx.run(query, doctor_id=doctor_id, name=name, specialty=specialty)
```

To create relationship between patient and doctor

```python
def create_treatment_relationship(tx, doctor_id, patient_id):
    query = """
    MATCH (d:Doctor {id: $doctor_id}), (p:Patient {id: $patient_id})
    CREATE (d)-[:TREATED]->(p)
    """
    tx.run(query, doctor_id=doctor_id, patient_id=patient_id)
```

To get the all relationships

```python
def get_treated_patients(tx):
    query = """
    MATCH (d:Doctor)-[:TREATED]->(p:Patient)
    RETURN d.name AS doctor, p.name AS patient
    """
    result = tx.run(query)
    for record in result:
        print(f"Doctor: {record['doctor']} treated  Patient: {record['patient']}")

        
```

**Store the appointment details** for easy access 

```python
def store_appointment(patient_id, doctor_id, appointment_date, appointment_time):
    # Store appointment
    appointment_key = f'appointment:{patient_id}'
    appointment_details = f'{doctor_id},{appointment_date},{appointment_time}'
    client.set(appointment_key, appointment_details)
    
    # Increment doctor visit count
    visit_count_key = f'visit_count:{doctor_id}'
    client.incr(visit_count_key)
    return print("details stored succefully")
```

**get the appointment details for specific doctor**

```python
def get_appointments_for_doctor(doctor_id):
    all_appointments = client.keys('appointment:*')
    doctor_appointments = []
    for appointment in all_appointments:
        details = client.get(appointment)
        doctor_id_from_details, date, time = details.split(',')
        if doctor_id_from_details == doctor_id: 
            patient_id = appointment.split(':')[1]  
            doctor_appointments.append({
                'patient_id': patient_id,
                'appointment_date': date,
                'appointment_time': time
            })
    return doctor_appointments
```

```python
def get_doctor_visits(doctor_id):
    return client.get(f'visit_count:{doctor_id}') or 0
```
